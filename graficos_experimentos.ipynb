```python
# Graficas para los experimentos

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from practica_performance import run_training
```

```python
# Experimento 1
ITERATIONS = 10

learning_rates = np.array([0.5, 1.5, 2.5, 3.5])            # Resultados
iterations = np.linspace(1, ITERATIONS, num=ITERATIONS)

combinaciones = np.array(np.meshgrid(iterations, learning_rates)).T.reshape(-1,2)
df_hiperparametros = pd.DataFrame(combinaciones, columns=['Iterations', 'Learning Rate'])


for i in range(len(df_hiperparametros)):
    print(f"Iterations: {df_hiperparametros["Iterations"][i]},  Learning Rate: {df_hiperparametros["Learning Rate"][i]}")
    df_hiperparametros["Cost"] = df_hiperparametros.index    # sustituir con el resultado de la función de coste después del entrenamiento
```

```python
for learning_rate, group in df_hiperparametros.groupby('Learning Rate'):
    plt.plot(group['Iterations'], group['Cost'], '-o', label=f"Learning rate = {learning_rate}")

plt.xlabel('Iteraciones')
plt.ylabel('Cost function')
plt.title('Convergencia de la función de coste')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Experimento 2
N_WORKERS = 4

execution_time = np.array([250, 150, 100, 50])            # Resultados
n_workers = np.linspace(1, N_WORKERS, num=N_WORKERS)


plt.plot(n_workers, execution_time, '-o')
plt.xticks(n_workers)
plt.title('Experimento 2 - Execution Time vs Workers')
plt.xlabel('Workers')
plt.ylabel('Execution time (s)')
plt.grid(True)
plt.show()


speed_up = execution_time[0] / execution_time
plt.plot(n_workers, speed_up, '-o')
plt.xticks(n_workers)
plt.title('Experimento 2 - Speed-up vs Workers')
plt.xlabel('Workers')
plt.ylabel('Speed-up (T1 / Tn)')
plt.grid(True)
plt.show()
```
