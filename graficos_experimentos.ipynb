```python
# Graficas para los experimentos

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from practica_performance import run_training
```

```python
# Experimento 1
ITERATIONS = 10

learning_rates = np.array([0.5, 1.5, 2.5, 3.5])            # Resultados
iterations = np.linspace(1, ITERATIONS, num=ITERATIONS)

combinaciones = np.array(np.meshgrid(iterations, learning_rates)).T.reshape(-1,2)
df_hiperparametros = pd.DataFrame(combinaciones, columns=['Iterations', 'Learning Rate'])


for i in range(len(df_hiperparametros)):
    print(f"Iterations: {df_hiperparametros["Iterations"][i]},  Learning Rate: {df_hiperparametros["Learning Rate"][i]}")
    df_hiperparametros["Cost"] = df_hiperparametros.index    # sustituir con el resultado de la función de coste después del entrenamiento
```

```python
for learning_rate, group in df_hiperparametros.groupby('Learning Rate'):
    plt.plot(group['Iterations'], group['Cost'], '-o', label=f"Learning rate = {learning_rate}")

plt.xlabel('Iteraciones')
plt.ylabel('Cost function')
plt.title('Convergencia de la función de coste')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Experimento 2
N_WORKERS = 4

execution_time = np.array([run_training(workers = 1)[0], run_training(workers = 2)[0], run_training(workers = 3)[0], run_training(workers = 4)[0]]) # Resultados
n_workers = np.linspace(1, N_WORKERS, num=N_WORKERS)


plt.plot(n_workers, execution_time, '-o')
plt.xticks(n_workers)
plt.title('Experimento 2 - Execution Time vs Workers')
plt.xlabel('Workers')
plt.ylabel('Execution time (s)')
plt.grid(True)
plt.show()


speed_up = execution_time[0] / execution_time
plt.plot(n_workers, speed_up, '-o')
plt.xticks(n_workers)
plt.title('Experimento 2 - Speed-up vs Workers')
plt.xlabel('Workers')
plt.ylabel('Speed-up (T1 / Tn)')
plt.grid(True)
plt.show()
```

```python
# Experimento 3

resultados = {
  "No cache" : run_training()[0],
  "Cache after map" : run_training(cacheMode="map")[0],
  "Cache at source" : run_training(cacheMode="source")[0],
  "Optimal mode" : run_training(cacheMode="opt")[0]
}

plt.bar(resultados.keys(), resultados.values())
plt.xticks(rotation=45)
plt.title('Experimento 3 - Catching strategy vs. Execution time')
plt.ylabel('Execution time (s)')
plt.show()

# 3.b => (Reptit experimetn 2 with the best caching strategy)

execution_time = np.array([run_training(workers = 1)[0], run_training(workers = 2)[0], run_training(workers = 3)[0], run_training(workers = 4)[0]]) # Resultados
execution_time_cached = np.array([run_training(workers = 1,cacheMode="opt")[0], run_training(workers = 2,cacheMode="opt")[0], run_training(workers = 3,cacheMode="opt")[0], run_training(workers = 4,cacheMode="opt")[0]]) # Resultados con estrategia óptima de caché
n_workers = np.linspace(1, N_WORKERS, num=N_WORKERS)


plt.plot(n_workers, execution_time, '-o', label='Sin caché')
plt.plot(n_workers, execution_time_cached, '-o', label='Con caché óptimo')

plt.xticks(n_workers)
plt.title('Experimento 2 - Execution Time vs Workers')
plt.xlabel('Workers')
plt.ylabel('Execution time (s)')
plt.grid(True)
plt.show()


speed_up = execution_time[0] / execution_time
speed_up_cached = execution_time[0] / execution_time_cached

plt.plot(n_workers, speed_up, '-o', label='Sin caché')
plt.plot(n_workers, speed_up_cached, '-o', label='Con caché óptimo')
plt.xticks(n_workers)
plt.title('Experimento 2 - Speed-up vs Workers')
plt.xlabel('Workers')
plt.ylabel('Speed-up (T1 / Tn)')
plt.grid(True)
plt.show()
```

```python
# Experimento 4
N_WORKERS = 4
N_PARTITIONS = 8

n_partitions = np.linspace(1, N_PARTITIONS, num=N_PARTITIONS)
execution_time = np.array([])
part = list(range(1, N_PARTITIONS + 1))  

for e in part:
    time, _ = run_training(N_WORKERS, e)
    execution_time = np.append(execution_time, time)

plt.plot(n_partitions, execution_time, '-o')
plt.xticks(n_partitions)
plt.legend(['Workers = 4'])
plt.title('Experimento 4 - Partitions')
plt.xlabel('Partitions')
plt.ylabel('Execution time (s)')
plt.grid(True)
plt.show()
```
